```python
nums = list(range(1, 11))
X = torch.tensor(nums, dtype=torch.float32)
Y = torch.tensor([2*x for x in nums], dtype=torch.float32)

w = torch.tensor(0.0, dtype=torch.float32, requires_grad=True)

def forward(x):
    return w * x

def loss(y, y_pred):
    return ((y_pred - y)**2).mean()

#forward(X_test).item()

learning_rate = 0.01
n_epoch = 10

for epoch in range(n_epoch):
    y_pred = forward(X)
    l = loss(Y, y_pred)
    l.backward()

    with torch.no_grad():
        w -= learning_rate * w.grad
    w.grad.zero_()

    print(f"epoch {epoch + 1}: w={w.item():.3f}, loss={l.item():.3f}")

for n in nums:
    print(f"f({n}) = {forward(n).item():.3f}")
    
```

```python
import torch
import torch.nn as nn

nums = [[n] for n in range(1, 11)]
nums2 = [[2*n] for n in range(1, 11)]
X = torch.tensor(nums, dtype=torch.float32)
Y = torch.tensor(nums2, dtype=torch.float32)

n_samples, n_features = X.shape

class LinearRegression(nn.Module):
    def __init__(self, input_dim, output_dim):
        super().__init__()
        self._lin = nn.Linear(input_dim, output_dim)

    def forward(self, x):
        return self._lin(x)

input_size, output_size = n_features, n_features

model = LinearRegression(input_size, output_size)

learning_rate = 0.01
n_epoch = 100

loss = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

for epoch in range(n_epoch):
    y_pred = model(X)
    l = loss(Y, y_pred)
    l.backward()
    optimizer.step()
    optimizer.zero_grad()

    if (epoch+1) % 10 == 0:
        w, b = model.parameters()
        print(f"epoch {epoch+1}: {w=}, {l=}")

for n in nums:
    print(f"f({n}) = {model(n).item():.3f}")
```
